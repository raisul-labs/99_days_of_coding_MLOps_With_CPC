```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
#extract sample feature for scaling comparison
```

```python

sample_features = data[['gender', 'age']].copy()

# 1. Standardization
standard_scaler = StandardScaler()

standard_scaled = pd.DataFrame(
    standard_scaler.fit_transform(sample_features),
    columns=['gender', 'age']
)

# 2. Normalization
minmax_scaler = MinMaxScaler()

minmax_scaled = pd.DataFrame(
    minmax_scaler.fit_transform(sample_features),
    columns=['gender', 'age']
)

print("Original Data:")
print(sample_features)

print("\nStandardized Data:")
print(standard_scaled)

print("\nNormalized Data:")
print(minmax_scaled)
```

```python
# Side by side comparison

fig, axes = plt.subplots(2, 3, figsize=(16, 8))

# Raw features
sns.histplot(sample_features['gender'], ax=axes[0, 0])
axes[0, 0].set_title("Raw Gender")

sns.histplot(sample_features['age'], ax=axes[1, 0])
axes[1, 0].set_title("Raw Age")


# StandardScaler
sns.histplot(standard_scaled['gender'], ax=axes[0, 1])
axes[0, 1].set_title("Standardized Gender")

sns.histplot(standard_scaled['age'], ax=axes[1, 1])
axes[1, 1].set_title("Standardized Age")


# MinMaxScaler
sns.histplot(minmax_scaled['gender'], ax=axes[0, 2])
axes[0, 2].set_title("MinMax Gender")

sns.histplot(minmax_scaled['age'], ax=axes[1, 2])
axes[1, 2].set_title("MinMax Age")


plt.tight_layout()
plt.show()
```
