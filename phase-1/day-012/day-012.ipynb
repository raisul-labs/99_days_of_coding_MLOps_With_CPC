```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = sns.load_dataset('diamonds')

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.hist(figsize=(15, 10), bins=20)
plt.suptitle("Feature Distributions")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Diamonds Dataset Correlation")
plt.show()
```

```python
df.columns
```

```python
columns = [
    'carat',
    'depth',
    'table',
    'x'
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for col, ax in zip(columns, axes.flatten()):
    sns.regplot(
        data=df,
        x=col,
        y='price',
        ax=ax,
        scatter_kws={'alpha': 0.4},
        line_kws={'color': 'red'}
    )

    ax.set_title(f'{col} vs Price')
    ax.set_xlabel(col)
    ax.set_ylabel('Price')

plt.tight_layout()
plt.show()
```
